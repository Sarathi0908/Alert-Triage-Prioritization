# 04 · Operating point & priority (Alert Triage #01)
The threshold is *fitted*, not fixed at 0.5: the analyst queue is precision-constrained, so we hold a precision floor and maximise recall under it. Priority then orders the queue — over PREDICTED ALERTS ONLY.

Logic: `src/features/priority_score_builder.PriorityScoreBuilder`.
